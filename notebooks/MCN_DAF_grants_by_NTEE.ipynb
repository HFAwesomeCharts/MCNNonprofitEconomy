### Clean and format columns in NTEE data

```python
ntee_raw_df.columns
```

```python
# Clean & format columns in NTEE data

ntee_clean_df = ntee_raw_df.copy()

# Convert integer columns to integer format
int_cols = ["EIN", "SUBSECTION_CODE"]
ntee_clean_df[int_cols] = ntee_clean_df[int_cols].astype("Int64")

# Drop unnecessary columns
cols_to_drop = ["ORG_ADDR_STATE", "NTEE_OLD_BMF", "NTEE_IRS", "NTEE_CODE_RAW", "NTEE_CODE_CLEAN", "NTEE_SOURCE"]
ntee_clean_df = ntee_clean_df.drop(columns=cols_to_drop)

# Uppercase NTEE_COMBINED
ntee_clean_df["NTEE_COMBINED"] = ntee_clean_df["NTEE_COMBINED"].str.upper()

# Create one-digit NTEE code columm
ntee_clean_df["NTEE_FIRST_DIGIT"] = ntee_clean_df["NTEE_COMBINED"].str[0]

# Create two-digit NTEE code columm
ntee_clean_df["NTEE_FIRST_THREE_DIGITS"] = ntee_clean_df["NTEE_COMBINED"].str[:3]

# Identify invalid first digits (not A-Z or null/NaN) and replace with "0"
ntee_clean_df["NTEE_FIRST_DIGIT"] = ntee_clean_df["NTEE_FIRST_DIGIT"].mask(
    ntee_clean_df["NTEE_FIRST_DIGIT"].isna() | ~ntee_clean_df["NTEE_FIRST_DIGIT"].str.match(r"^[A-Z]$", na=False),
    "0"
)
ntee_clean_df["NTEE_FIRST_THREE_DIGITS"] = ntee_clean_df["NTEE_FIRST_THREE_DIGITS"].mask(
    ntee_clean_df["NTEE_FIRST_THREE_DIGITS"].isna()
    | ~ntee_clean_df["NTEE_FIRST_THREE_DIGITS"].str.match(r"^[A-Z][0-9][0-9]$", na=False),
    "0"
)

ntee_clean_df.head()
```

```python
# Find records where NTEE_COMBINED is null or 0

# Rows where NTEE_FIRST_DIGIT is NaN or not A-Z
invalid_ntee_rows = ntee_clean_df[
    ntee_clean_df["NTEE_FIRST_DIGIT"].isna() |        # null/NaN
    ~ntee_clean_df["NTEE_FIRST_DIGIT"].str.match(r"^[A-Z]$", na=False)  # not a single uppercase letter
]

# print(invalid_ntee_rows.value_counts())
# Show the results
# print(invalid_ntee_rows)
print("Number of invalid NTEE codes:", len(invalid_ntee_rows))
```

```python
# Dedupe the NTEE file

ntee_valid_df = ntee_clean_df.copy()
print("Before deduping:", len(ntee_valid_df))

# Find rows with valid NTEE codes
ntee_valid_df["is_valid_ntee"] = (
    ntee_valid_df["NTEE_FIRST_DIGIT"]
    .astype(str)
    .str.match("^[A-Za-z]$")
)

# Make sure the year is numeric
ntee_valid_df["LAST_YEAR_IN_BMF"] = pd.to_numeric(ntee_valid_df["LAST_YEAR_IN_BMF"], errors="coerce")

# Sort by EIN, valid ntee code flag, and LAST_YEAR_IN_BMF
ntee_sorted_df = ntee_valid_df.sort_values(
    by=["EIN", "is_valid_ntee", "LAST_YEAR_IN_BMF"],
    ascending=[True, False, False]
)

# Keep only the first record for each EIN
ntee_dedupe_df = ntee_sorted_df.drop_duplicates(subset="EIN", keep="first")

# Drop the helper column
ntee_dedupe_df = ntee_dedupe_df.drop(columns=["is_valid_ntee"])

print("After deduping:", len(ntee_dedupe_df))

ntee_dedupe_df.head()
```

### Remove DAF-to-DAF grants from All-DAF grants (for no DAF-to-DAF file)

```python
# Remove DAF2DAF grants from All-DAF grants list

print("Before filtering:", len(alldaf_allntee_df))
print("Records in DAF2DAF list:", len(daf2daf_clean_df))

merged = (
    alldaf_allntee_df
    .merge(
        daf2daf_clean_df[["Grantor EIN", "Recipient EIN"]],
        left_on=["EIN", "RcpntTbl_RcpntEIN"],
        right_on=["Grantor EIN", "Recipient EIN"],
        how="left",
        indicator=True
    )
)

# Keep only rows that are NOT daf-to-daf grants
no_daf2daf_filtered_df = merged[merged["_merge"] == "left_only"].copy()

# Keep only rows that ARE daf-to-daf grants
daf2daf_filtered_df = merged[merged["_merge"] == "both"].copy()

# Drop helper columns
no_daf2daf_filtered_df = no_daf2daf_filtered_df.drop(columns=["Grantor EIN", "Recipient EIN", "_merge"])
daf2daf_filtered_df = daf2daf_filtered_df.drop(columns=["Grantor EIN", "Recipient EIN", "_merge"])

print("All non DAF2DAF:", len(no_daf2daf_filtered_df))
print("DAF2DAF only:", len(daf2daf_filtered_df))

print("Sanity check total:", len(no_daf2daf_filtered_df) + len(daf2daf_filtered_df))

no_daf2daf_filtered_df.head()
```

### Create summaries by NTEE code

```python
# Function to calculate Shonni's summary groupings on 

def shonni_groups(row, first_digit_ntee_code, full_ntee_code):

    code = row[first_digit_ntee_code]
    combined = str(row[full_ntee_code]).upper() if pd.notna(row[full_ntee_code]) else ""
    
    if pd.isna(code):
        return None
    
    code = str(code).upper()

    if combined.startswith(("T30", "T31", "T32")): # private foundations, community foundations, and corporate foundations
        return "Foundations"
    
    if code == "A":
        return "Arts, Culture, Humanities"
    
    elif code == "B":
        if combined.startswith(("B40", "B41", "B43", "B50")):
            return "Colleges and Universities"
        else:
            return "Other Education"
    
    elif code in ["C", "D"]:
        return "Environment and Animals"
    
    elif code in ["E", "F", "G", "H"]:
        if combined.startswith(("E20", "E21", "E22", "E24", "E90", "E91", "E92")):
            return "Hospitals and Nursing Homes"
        else:
            return "Other Health"
    
    elif code in ["I", "J", "K", "L", "M", "N", "O", "P"]:
        return "Human Services"
    
    elif code == "Q":
        return "International, Foreign Affairs"
    
    elif code in ["R", "S", "T", "U", "V", "W"]:
        return "Public, Societal Benefit"
    
    elif code == "X":
        return "Religion Related"
    

    elif code in ["Y", "Z"]:
        return "Unknown, Unclassified"
    
    else:
        return None
```

```python
# Apply Shonni groupings to daf-to-daf grants by RECIPIENT NTEE code

d2d_recip_summary_df = daf2daf_filtered_df.copy()

d2d_recip_summary_df["NTEE_ACTIVITY"] = d2d_recip_summary_df.apply(
    shonni_groups,
    axis=1,
    args=("RECIPIENT_NTEE_FIRST_DIGIT", "RECIPIENT_NTEE_COMBINED")
)

d2d_recip_summary_df.head()
```

```python
# Apply Shonni groupings to daf-to-daf grants by RECIPIENT NTEE code

no_d2d_recip_summary_df = no_daf2daf_filtered_df.copy()

no_d2d_recip_summary_df["NTEE_ACTIVITY"] = no_d2d_recip_summary_df.apply(
    shonni_groups,
    axis=1,
    args=("RECIPIENT_NTEE_FIRST_DIGIT", "RECIPIENT_NTEE_COMBINED")
)

no_d2d_recip_summary_df.head()
```

```python
# Define activity area sort order

activity_order = ["Arts, Culture, Humanities", "Other Education", "Colleges and Universities", "Environment and Animals", "Other Health", 
         "Hospitals and Nursing Homes", "Human Services", "International, Foreign Affairs", "Public, Societal Benefit",
         "Foundations", "Religion Related", "Unknown, Unclassified"]
```

```python
# Create summary of daf-to-daf data

d2d_recip_summary_df = daf2daf_filtered_df.copy()

d2d_recip_summary_df["NTEE_ACTIVITY"] = d2d_recip_summary_df.apply(
    shonni_groups,
    axis=1,
    args=("RECIPIENT_NTEE_FIRST_DIGIT", "RECIPIENT_NTEE_COMBINED")
)

d2d_recip_summary = (
    d2d_recip_summary_df
    .groupby("NTEE_ACTIVITY", as_index=False)
    .agg(
        Count_of_Grants=("RcpntTbl_CshGrntAmt", "count"),
        Grant_Amount=("RcpntTbl_CshGrntAmt", "sum")
    )
)

d2d_recip_summary["NTEE_ACTIVITY"] = pd.Categorical(
    d2d_recip_summary["NTEE_ACTIVITY"],
    categories=activity_order,
    ordered=True
)

d2d_recip_summary = d2d_recip_summary.sort_values("NTEE_ACTIVITY")

d2d_recip_summary
```

```python
# Create summary of no-daf-to-daf data

no_d2d_recip_summary_df = no_daf2daf_filtered_df.copy()

no_d2d_recip_summary_df["NTEE_ACTIVITY"] = no_d2d_recip_summary_df.apply(
    shonni_groups,
    axis=1,
    args=("RECIPIENT_NTEE_FIRST_DIGIT", "RECIPIENT_NTEE_COMBINED")
)

no_d2d_recip_summary = (
    no_d2d_recip_summary_df
    .groupby("NTEE_ACTIVITY", as_index=False)
    .agg(
        Count_of_Grants=("RcpntTbl_CshGrntAmt", "count"),
        Grant_Amount=("RcpntTbl_CshGrntAmt", "sum")
    )
)

no_d2d_recip_summary["NTEE_ACTIVITY"] = pd.Categorical(
    no_d2d_recip_summary["NTEE_ACTIVITY"],
    categories=activity_order,
    ordered=True
)

no_d2d_recip_summary = no_d2d_recip_summary.sort_values("NTEE_ACTIVITY")

no_d2d_recip_summary
```

```python
# Merge the no-daf-to-daf and the daf-to-daf data together into one summary table by RECIPIENT NTEE code

# Rename the columns in each summary dataframe
d2d_summary = d2d_recip_summary.rename(columns={
    "Count_of_Grants": "Count_D2D",
    "Grant_Amount": "Grant_Amount_D2D"
})

no_d2d_summary = no_d2d_recip_summary.rename(columns={
    "Count_of_Grants": "Count_No_D2D",
    "Grant_Amount": "Grant_Amount_No_D2D"
})

# Merge the two

combined_recipient_summary = d2d_summary.merge(
    no_d2d_summary,
    on="NTEE_ACTIVITY",
    how="outer"
)

# Fill any NaN numeric records with 0
combined_recipient_summary[[
    "Count_D2D",
    "Grant_Amount_D2D",
    "Count_No_D2D",
    "Grant_Amount_No_D2D"
]] = combined_recipient_summary[[
    "Count_D2D",
    "Grant_Amount_D2D",
    "Count_No_D2D",
    "Grant_Amount_No_D2D"
]].fillna(0)

combined_recipient_summary
```
